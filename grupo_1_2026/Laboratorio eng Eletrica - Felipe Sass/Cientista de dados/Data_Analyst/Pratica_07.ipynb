## Detecção de linhas de plantio com Hough Line Transform

image.png

A transformada de Hough para detecção de linhas é uma técnica popular para identificar linhas em imagens, incluindo imagens obtidas por drones. A transformada de Hough converte pontos na imagem para uma representação paramétrica de linhas, permitindo a detecção de linhas mesmo em imagens com ruído ou descontinuidades.

***Etapas***

- Pré-processamento da imagem: Antes de aplicar a transformada de Hough, é útil realizar algumas etapas de pré-processamento na imagem para melhorar a qualidade dos resultados. Isso pode incluir a aplicação de filtros para reduzir o ruído, como um filtro de suavização (como filtro gaussiano) ou um filtro de detecção de bordas (como Canny).

- Detecção de bordas: A transformada de Hough opera em bordas na imagem, então é necessário aplicar um algoritmo de detecção de bordas na imagem pré-processada. O algoritmo de Canny é amplamente utilizado para essa finalidade, pois é eficaz na detecção de bordas e minimiza a resposta a ruídos.

- Aplicação da transformada de Hough: A transformada de Hough é aplicada na imagem de bordas para identificar os parâmetros (ângulo e distância) das linhas presentes na imagem. A transformada de Hough padrão opera no espaço de parâmetros de ângulo e distância (ou pode ser estendida para detectar linhas com parâmetros adicionais, como espessura).

- Limiarização e detecção de linhas: Após a aplicação da transformada de Hough, você precisará definir um limiar para identificar as linhas significativas na imagem. Os pontos na transformada de Hough que excedem o limiar são considerados como representantes das linhas detectadas. Você pode ajustar o limiar para controlar a sensibilidade da detecção.

- Desenhar as linhas detectadas na imagem original: Por fim, com base nos parâmetros das linhas detectadas na etapa anterior, você pode desenhar as linhas encontradas na imagem original. Isso permitirá que você visualize as linhas de plantio identificadas.

É importante mencionar que a transformada de Hough pode ser sensível a diferentes condições de iluminação, texturas e outros fatores presentes nas imagens. Portanto, pode ser necessário ajustar os parâmetros da transformada de Hough e realizar testes em diferentes cenários para obter os melhores resultados.

Além disso, considere que a detecção de linhas de plantio em imagens de drones pode envolver desafios específicos, como a presença de vegetação, variações de iluminação e sobreposição de linhas. Portanto, pode ser necessário explorar técnicas adicionais, como segmentação de imagem ou filtros personalizados, para lidar com essas situações e melhorar a precisão da detecção de linhas de plantio.

Instalamos as bibliotecas que vamos usar

```python
import geopandas as gpd
import numpy as np
import pandas as pd
import shapely
import math
import os
import rasterio
from rasterio.mask import mask
import json
from rasterio.plot import show
import matplotlib.pyplot as plt
import cv2
from skimage.morphology import skeletonize, convex_hull_image
from skimage.measure import label, regionprops
from shapely.geometry import shape
import shapely
from rasterio.features import shapes
from shapely.geometry import LineString
```

Então setamos o caminho da imagem que vamos usar:

```python
path_img = '/content/drive/MyDrive/Datasets/Lines_planting/AOI_img.tif'
```

Com o rasterio abrimos a imagem e plotamos com o matplotlib:

```python
src = rasterio.open(path_img)
```

```python
img = src.read()
```

```python
img = img.transpose([1,2,0])
```

```python
img = img.astype('uint8')
```

```python
plt.figure(figsize=[16,16])
plt.imshow(img)
```

Na nossa imagem RGB vamos separar as bandas espectrais e selecionar a primeira banda espectral.

```python
band = img[:,:,0]
```

```python
plt.figure(figsize=[16,16])
plt.imshow(band,cmap='RdYlGn')
plt.axis('off')
```

Vamos ver a distribuição do histograma dessa banda:

```python
pixels = img[:,:,0].ravel()
pixels = pixels[pixels != 0]
```

```python
ax = plt.hist(pixels, bins = 256)
plt.show()
```

```python
band  = img[:,:,0].copy()
band = np.where(band == 0 , 255, band)
```

Selecionamos um valor para dividir a distribuição em duas classes:

```python
_,thresh1 = cv2.threshold(band,90,1,cv2.THRESH_BINARY_INV)
```

```python
plt.figure(figsize=[16,16])
plt.imshow(thresh1, cmap='Greys_r')
```

Agora vamos obter o polígono envolvente da nossa região:

```python
hull = convex_hull_image(thresh1 == 1)
```

```python
plt.figure(figsize=[16,16])
plt.imshow(hull.astype('uint8'), cmap='Greys_r')
```

E converter a imagem do polígono em um geodataframe que usaremos depois.

```python
shape_gen = ((shape(s), v) for s, v in shapes(hull.astype('uint8'), mask=hull, transform=src.transform))
Poly_gdf = gpd.GeoDataFrame(dict(zip(["geometry", "class"], zip(*shape_gen))), crs=src.crs)
```

Ainda com a imagem do poligono, vamos obter a orientação desse poligono:

```python
for region in regionprops(hull.astype('uint8')):
  orientation = region.orientation
  angle_in_degrees = orientation * (180/np.pi)
  print(angle_in_degrees)
```

Então aplicaremos uma rotação na imagem RGB usando o angulo que obtemos:

```python
rows,cols,_ = img.shape
```

```python
M = cv2.getRotationMatrix2D(((cols-1)/2.0,(rows-1)/2.0),-32,1)
dst_img = cv2.warpAffine(img,M,(cols,rows))
```

```python
plt.figure(figsize=[16,16])
plt.imshow(dst_img)
plt.axis('off')
```

Também aplicaremos a rotação na imagem binária:

```python
rows,cols, = thresh1.shape
```

```python
M = cv2.getRotationMatrix2D(((cols-1)/2.0,(rows-1)/2.0),-32,1)
dst = cv2.warpAffine(thresh1,M,(cols,rows))
```

```python
plt.figure(figsize=[16,16])
plt.imshow(dst)
plt.axis('off')
```

Vamos agora aplicar os elementos morfológicos de abertura e dilatação usando kernels retangulares para filtrar as linhas na imagem:

```python
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 25))
morph = cv2.morphologyEx(dst, cv2.MORPH_OPEN, kernel)

kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (50, 5))
morph = cv2.morphologyEx(morph, cv2.MORPH_DILATE, kernel)

# use morphology open to remove thin lines from dotted lines
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (10,3))
morph = cv2.erode(morph, kernel,  iterations = 5)
```

```python
plt.figure(figsize=[16,16])
plt.imshow(morph)
plt.axis('off')
```

Aplicamos a esqueletonização para obter linhas finas:

```python
from skimage.morphology import skeletonize, convex_hull_image
```

```python
sk = skeletonize(morph == 1)
```

```python
plt.figure(figsize=[16,16])
plt.imshow(sk,cmap='gray')
plt.axis('off')
```

Como essas linhas estão com muitos ruídos verticais, vamos filtrar a imagem com um sobel na direção horizontal:

```python
sobely = cv2.Sobel(morph.astype('uint8'),cv2.CV_8U,0,1,ksize=5)
```

```python
plt.figure(figsize=[16,16])
plt.imshow(sobely,cmap='gray')
plt.axis('off')
```

Com isso podemos retornar a imagem à posição original.

```python
rows,cols,_ = img.shape
```

```python
M = cv2.getRotationMatrix2D(((cols-1)/2.0,(rows-1)/2.0),32,1)
result_img = cv2.warpAffine(sobely,M,(cols,rows))
```

```python
plt.figure(figsize=[16,16])
plt.imshow(result_img)
plt.axis('off')
```

Podemos usar o Hough Lines para identificar as linha de plantio:

```python
from skimage.transform import hough_line, hough_line_peaks
from skimage.feature import canny
from skimage.draw import line
```

```python
tested_angles = np.linspace(-np.pi / 2, np.pi / 2, 360)

h, theta, d = hough_line(result_img, theta=tested_angles)
hpeaks = hough_line_peaks(h, theta, d, threshold=0.50 * h.max())
```

```python
for _, angle, dist in zip(*hpeaks):
  print(dist)
```

Vamos criar um geodataframe de linhas e plotá-las junto com a imagem RGB:

```python
plt.figure(figsize=(16,16))
plt.imshow(img)
i = 0
crs_target = src.crs.to_dict()['init']
gdf_full = gpd.GeoDataFrame([])
lines = []
lines_id = []
row1, col1, _ = img.shape
for _, angle, dist in zip(*hpeaks):
    y0 = (dist - 0 * np.cos(angle)) / np.sin(angle)
    y1 = (dist - col1 * np.cos(angle)) / np.sin(angle)
    pt1 = (0,int(y0))
    pt2 = (col1,int(y1))
    retval, pt1_result, pt2_result = cv2.clipLine((0,0,col1,row1), pt1, pt2)
    plt.plot((pt1_result[0], pt2_result[0]), (pt1_result[1], pt2_result[1]), '-r')

    coordinates_pt1 = rasterio.transform.xy(src.transform, pt1_result[1], pt1_result[0])
    coordinates_pt2 = rasterio.transform.xy(src.transform, pt2_result[1], pt2_result[0])

    coords = [coordinates_pt1,coordinates_pt2]
    line = LineString(coords)
    lines.append(line)
    lines_id.append(i)
    i = i+1
gdf = gpd.GeoDataFrame(data=lines_id,geometry=lines)
gdf = gdf.rename(columns={0:"Line_Id"})
gdf = gdf.set_crs(crs_target)
gdf_full = pd.concat([gdf_full,gdf])
```

Ao observar o geodataframe de linhas, vimos que as linhas saem fora da área válida da imagem:

```python
gdf_full.plot()
```

Vamos usar o geodataframe do poligono envolvente para cortar as partes das linhas que não estão na área valida da imagem:

```python
Lines_cliped = gpd.overlay(gdf_full, Poly_gdf, how='intersection')
```

```python
Lines_cliped.plot()
```
